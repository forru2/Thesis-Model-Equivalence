## Features selection

```python
def get_pca_metrics(df, family = None, n_components = 1):
    pca = PCA(n_components = n_components)
    
    performance = pca.fit_transform(df[['val_accuracy', 'val_f1', 'val_precision', 'val_recall']])
    fairness = pca.fit_transform(df[['val_equalized_odds_inverted', 'val_cuae_inverted','val_dem_parity_inverted', 'val_equal_opportunity_inverted',
                                     'val_predictive_equality_inverted', 'val_predictive_parity_inverted', 'val_neg_predictive_parity_inverted']])
    fair_robustness = pca.fit_transform(df[['val_dem_parity_robustness', 'val_equal_opp_robustness', 'val_predictive_eq_robustness',
                                                           'val_pred_parity_robustness', 'val_neg_pred_parity_robustness',
                                                           'val_eq_odds_robustness']])
    
    perf_robustness = pca.fit_transform(df[['val_acc_robustness', 'val_f1_robustness', 'val_precision_robustness', 'val_recall_robustness']])
    
    if family == 'rtc':
        complexity = pca.fit_transform(df[['val_rtc_depth', 'val_rtc_n_nodes', 'val_rtc_n_rules']])[:,0]
    elif family == 'knn':
        complexity = df['knn_complexity'].values
    else:
        complexity = df['val_coeff_magnitude'].values

    d = {
        'performance': performance[:,0], 
        'fairness': fairness[:,0], 
        'fair_robustness': fair_robustness[:,0], 
        'perf_robustness': perf_robustness[:,0], 
        'complexity': complexity
    }

    return pd.DataFrame(d)
```

```python
rtc_pca = get_pca_metrics(scaled_rtc, family='rtc')
knn_pca = get_pca_metrics(scaled_knn, family='knn')
lr_pca = get_pca_metrics(scaled_lr, family='lr')

full_pca = get_pca_metrics(scaled_df)
```

## Rashomon Set Study

```python
def rr_variation(df, threshold, tol, metric):
    ratios = []
    for t in tol:
        cond = f'{metric} >= {threshold - (threshold*t)}'
        ratio = rashomon_ratio(df, cond)        
        ratios.append(ratio)
    return ratios
```

## Equivalence Study for the RS

### Similarity matrices computation

```python
def load_models(model_ids, model_type = None, path = '.'):
    models_dict = {}
    for model_id in model_ids:
        search_pattern = f"**/{model_id}.joblib"
        files_found = list(Path(path).rglob(search_pattern))
        if files_found:
            models_dict[model_id] = load(files_found[0])
    return models_dict

def get_preds_dict(models_dict, X_val):
    preds_dict = {}
    for model_id, model in models_dict.items():
        preds,_ = predictions(model, X_val)
        preds_dict[model_id] = preds
    return preds_dict

def get_metrics_dict(metric_col: str, df, key_col = 'model_id'):
    return df.set_index(key_col)[metric_col].to_dict()


def compute_distance(v1, v2, i = 0, j = 0, metric = hamming_loss):
    score = apply_distance_metric(v1, v2, metric)
    return i, j, score

def get_score_matrix(metrics_dict, metric = hamming_loss, n_jobs = -1):
    ids = list(metrics_dict.keys())
    metrics_list = [np.atleast_1d(metrics_dict[k]) for k in ids]

    if len(metrics_list[0]) == 1:
        values = np.array(metrics_list).reshape(-1, 1)
        eq_matrix = np.abs(values - values.T)
    else:
        results = Parallel(n_jobs = n_jobs)(delayed(compute_distance)(metrics_list[i], metrics_list[j], i, j, metric = metric)
                              for i in range(len(metrics_list)) 
                              for j in range(i+1, len(metrics_list))
                             )
        eq_matrix = np.zeros((len(metrics_list), len(metrics_list)))
        for i, j, score in results:
            eq_matrix[i, j] = eq_matrix[j, i] = score
    return eq_matrix, ids
```

```python
df_path = r'C:\Users\franc\OneDrive\Magistrale\Thesis-Model-Equivalence\Split_salvati\german'
model_path = r'C:\Users\franc\OneDrive\Magistrale\Thesis-Model-Equivalence\Models\german'

X_val = np.genfromtxt(f'{df_path}\german_credit_X_val.csv', delimiter=',', skip_header=1)
models_dict = load_models(model_ids = rash_set, path = model_path)

#otteniamo i dizionari delle metriche e previsioni
fairness_cols = ['val_equalized_odds', 'val_dem_parity',
                 'val_predictive_parity', 'val_neg_predictive_parity']
fairness_metrics = {col: get_metrics_dict(metric_col = col, df = rash_df) for col in fairness_cols}

fair_rob_cols = ['val_dem_parity_robustness', 'val_equal_opp_robustness', 'val_pred_parity_robustness', 
                'val_neg_pred_parity_robustness', 'val_eq_odds_robustness']
fair_rob_metrics = {col: get_metrics_dict(metric_col = col, df = rash_df) for col in fair_rob_cols}

perf_rob_cols = ['val_acc_robustness', 'val_f1_robustness', 'val_precision_robustness', 'val_recall_robustness']
perf_rob_metrics = {col: get_metrics_dict(metric_col = col, df = rash_df) for col in perf_rob_cols}

preds_dict = get_preds_dict(models_dict, X_val)

#otteniamo le matrici di errore (differenza assoluta per valori scalari e hamming normalizzata per vettore previsioni)
fairness_error_matrices = {metric: {'matrix': get_score_matrix(metrics_dict = fairness_metrics[metric])[0], 
                                    'ids': get_score_matrix(metrics_dict = fairness_metrics[metric])[1]}
                          for metric in fairness_metrics.keys()}

fair_rob_error_matrices = {metric: {'matrix': get_score_matrix(metrics_dict = fair_rob_metrics[metric])[0], 
                                    'ids': get_score_matrix(metrics_dict = fair_rob_metrics[metric])[1]}
                          for metric in fair_rob_metrics.keys()}

perf_rob_error_matrices = {metric: {'matrix': get_score_matrix(metrics_dict = perf_rob_metrics[metric])[0], 
                                    'ids': get_score_matrix(metrics_dict = perf_rob_metrics[metric])[1]}
                          for metric in perf_rob_metrics.keys()}

preds_error_matrix, preds_id = get_score_matrix(metrics_dict = preds_dict)
```

```python
cluster_idxs = {l: np.where(labels == l)[0] for l in set(labels)}
cluster_models = {l: np.array(ids)[idxs] for l, idxs in cluster_idxs.items()}

#cluster_metrics = {l: rash_df[col][rash_df['model_id'].isin(vals)].values for l, vals in cluster_models.items()}

for cl, vals in cluster_models.items():
    models = rash_df['model_type'][rash_df['model_id'].isin(vals)].values
    unique, counts = np.unique(models, return_counts=True)
    counts_str = ', '.join([f'{u}: {c}' for u, c in zip(unique, counts)])
    print(f'{cl}: {len(vals)}, {counts_str}')
```

```python
cluster_avg_distances = {}

for label, idxs in cluster_idxs.items():
    # Se il cluster ha solo un modello, la distanza media è 0
    if len(idxs) < 2:
        cluster_avg_distances[label] = 0.0
        continue
    
    # Estraiamo la sottomatrice del cluster usando np.ix_
    # Questo seleziona le righe e le colonne corrispondenti agli indici del cluster
    sub_matrix = error_matrix[np.ix_(idxs, idxs)]
    
    # Dobbiamo calcolare la media dei valori escludendo la diagonale (che è 0)
    # Esistono due modi comuni:
    
    # 1. Media di tutte le coppie (i, j) con i != j
    # Numero di elementi fuori dalla diagonale: n * (n - 1)
    n = len(idxs)
    sum_distances = np.sum(sub_matrix) # La somma include la diagonale che è 0
    avg_dist = sum_distances / (n * (n - 1))
    
    cluster_avg_distances[label] = avg_dist

# Visualizzazione risultati
for label, dist in cluster_avg_distances.items():
    print(f"Cluster {label} ({len(cluster_models[label])} modelli): Distanza media interna = {dist:.4f}")
```
